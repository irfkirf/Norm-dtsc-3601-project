```python
# =========================
# 0) Setup
# =========================
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import precision_recall_fscore_support

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# File path to your cleaned dataset from Sprint 2
DATA_PATH = "cleaned_property_crimes.csv"  # change if needed
RANDOM_STATE = 42
```

```python
# =========================
# 1) Load data
# =========================
df = pd.read_csv(DATA_PATH)

# Expecting columns similar to your screenshots:
# 'crime_group', 'crime_binary', 'Vict Age Group', 'Vict Sex', 'Vict Descent',
# 'AREA', 'LAT', 'LON', 'Hour', 'Month', 'Day0fWeek', (optionally 'Premis Cd')

print("Rows, Cols:", df.shape)
df.head()
```

```python
# =========================
# 2) Define target + features
# =========================
# Binary target (from your notebook): 1 = Theft, 0 = Break-in/Burglary
target_col = "crime_binary"

# Choose low-cardinality, useful features you already explored
categorical_cols = [
    "Vict Age Group", "Vict Sex", "Vict Descent",
    "AREA", "Month", "Day0fWeek"
]
numeric_cols = ["Hour", "LAT", "LON"]

# Keep only rows with non-missing target
df = df.dropna(subset=[target_col])

# Ensure all chosen columns exist
available_cat = [c for c in categorical_cols if c in df.columns]
available_num = [c for c in numeric_cols if c in df.columns]
X = df[available_cat + available_num].copy()
y = df[target_col].astype(int)

print("Categorical features used:", available_cat)
print("Numeric features used:", available_num)
print("Target distribution (1=Theft, 0=Break/Burg):")
print(y.value_counts(normalize=True).round(3))
```

```python
# =========================
# 3) Train/Test split
# =========================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
X_train.shape, X_test.shape
```

```python
# =========================
# 4) Preprocessing
# =========================
# One-hot encode categoricals; standardize numerics for Logistic Regression.
# (Trees don't need scaling, but leaving numerics "passthrough" is fine in the same preprocessor.)

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), available_cat),
        ("num", "passthrough", available_num)
    ]
)
```

```python
# =========================
# 5) Define models + param grids
#    (keep grids small so they finish fast)
# =========================

# Logistic Regression
pipe_log = Pipeline(steps=[
    ("prep", preprocessor),
    ("clf", LogisticRegression(max_iter=500, class_weight="balanced", n_jobs=None))
])
grid_log = {
    "clf__C": [0.5, 1.0, 2.0],
    "clf__penalty": ["l2"],
    "clf__solver": ["lbfgs"]
}

# Decision Tree
pipe_dt = Pipeline(steps=[
    ("prep", preprocessor),
    ("clf", DecisionTreeClassifier(random_state=RANDOM_STATE, class_weight="balanced"))
])
grid_dt = {
    "clf__max_depth": [8, 12, 16, None],
    "clf__min_samples_split": [2, 10, 50],
    "clf__min_samples_leaf": [1, 5, 20]
}

# Random Forest
pipe_rf = Pipeline(steps=[
    ("prep", preprocessor),
    ("clf", RandomForestClassifier(random_state=RANDOM_STATE, class_weight="balanced"))
])
grid_rf = {
    "clf__n_estimators": [150, 250],
    "clf__max_depth": [12, 20, None],
    "clf__min_samples_split": [2, 10],
    "clf__min_samples_leaf": [1, 5]
}

models = [
    ("Logistic Regression", pipe_log, grid_log),
    ("Decision Tree",       pipe_dt,  grid_dt),
    ("Random Forest",       pipe_rf,  grid_rf),
]
```

```python
from sklearn.model_selection import train_test_split

# target_col already set to "crime_binary"
N = 60_000                          # pick your subset size
train_size = min(N, len(df))        # cap if dataset is smaller

# Stratified subset: returns two splits; we keep the subset and drop the rest
df_small, _ = train_test_split(
    df,
    train_size=train_size,
    stratify=df[target_col],
    random_state=RANDOM_STATE
)

# proceed using df_small instead of df
categorical_cols = [c for c in ["Vict Age Group","Vict Sex","Vict Descent","AREA","Month","Day0fWeek"] if c in df_small.columns]
numeric_cols     = [c for c in ["Hour","LAT","LON"] if c in df_small.columns]

X = df_small[categorical_cols + numeric_cols].copy()
y = df_small[target_col].astype(int)

```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

try:
    ohe = OneHotEncoder(
        handle_unknown="infrequent_if_exist",
        min_frequency=50,
        sparse_output=True   # sklearn >= 1.2
    )
except TypeError:
    ohe = OneHotEncoder(
        handle_unknown="infrequent_if_exist",
        min_frequency=50,
        sparse=True          # fallback for older sklearn
    )

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", ohe, categorical_cols),
        ("num", "passthrough", numeric_cols)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_recall_fscore_support
import pandas as pd

def eval_model(name, pipe):
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(y_test, y_pred, average="binary", zero_division=0)
    print(f"\n=== {name} ===")
    print("Test Accuracy:", round(acc, 4))
    print(classification_report(y_test, y_pred, zero_division=0))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
    return {"Model": name, "Test Accuracy": acc, "Precision": prec, "Recall": rec, "F1": f1}

# Models tuned for speed
pipe_log = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(max_iter=500, solver="lbfgs"))  # no class_weight to keep accuracy higher
])

pipe_dt = Pipeline([
    ("prep", preprocessor),
    ("clf", DecisionTreeClassifier(max_depth=10, min_samples_split=20, min_samples_leaf=8, random_state=RANDOM_STATE))
])

pipe_rf = Pipeline([
    ("prep", preprocessor),
    ("clf", RandomForestClassifier(
        n_estimators=80, max_depth=12, min_samples_split=20, min_samples_leaf=8,
        max_features="sqrt", n_jobs=-1, random_state=RANDOM_STATE
    ))
])

results = []
for name, pipe in [("Logistic Regression", pipe_log),
                   ("Decision Tree", pipe_dt),
                   ("Random Forest", pipe_rf)]:
    results.append(eval_model(name, pipe))

res_df = pd.DataFrame(results).sort_values("Test Accuracy", ascending=False).reset_index(drop=True)
res_df
```

```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_fscore_support

def breakin_recall(model, name):
    yhat = model.predict(X_test)
    # order = [0,1] where 0 = Break-in, 1 = Theft
    prec, rec, f1, _ = precision_recall_fscore_support(
        y_test, yhat, labels=[0,1], average=None, zero_division=0
    )
    return {"Model": name, "Recall_BreakIn": rec[0]}

rows = [
    {"Model": "Baseline", "Recall_BreakIn": 0.0},  # majority-class baseline
    breakin_recall(pipe_log, "Logistic Regression"),
    breakin_recall(pipe_dt,  "Decision Tree"),
    breakin_recall(pipe_rf,  "Random Forest"),
]
minority_df = pd.DataFrame(rows)

plt.figure(figsize=(6,4))
plt.bar(minority_df["Model"], minority_df["Recall_BreakIn"]*100)
plt.ylabel("Recall for Break-ins (%)")
plt.title("Minority-Class Recall by Model")
plt.ylim(0,100)
# add labels on bars
for i, v in enumerate(minority_df["Recall_BreakIn"]*100):
    plt.text(i, v + 1, f"{v:.1f}%", ha="center", va="bottom", fontsize=9)
plt.tight_layout()
# plt.savefig("breakin_recall_by_model.png", dpi=180)   # uncomment to save
plt.show()
```
