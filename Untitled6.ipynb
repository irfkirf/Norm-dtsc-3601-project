```python
!pip -q install pgeocode shap shapely

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, RocCurveDisplay)
from sklearn.inspection import permutation_importance
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
import pgeocode
import warnings; warnings.filterwarnings("ignore")
```

```python
LAPD_PATH = "cleaned_property_crimes.csv"
ACS_ZIP_PATH = "ACS_data.B19013-Data.csv"
df = pd.read_csv(LAPD_PATH)


df = df.dropna(subset=["LAT","LON"])
df["crime_group"] = df["crime_group"].astype(str)


if "DATE_OCC" in df.columns:
    dt = pd.to_datetime(df["DATE_OCC"], errors="coerce")
    df["hour"] = dt.dt.hour
    df["dow"] = dt.dt.dayofweek
else:
    for col in ["hour","dow"]:
        if col not in df.columns:
            df[col] = np.nan

cat_cols = [c for c in ["AREA NAME","PremisesDesc","WeaponDesc"] if c in df.columns]
num_cols = [c for c in ["hour","dow","LAT","LON"] if c in df.columns]

target = "crime_group"
X_base = df[cat_cols + num_cols].copy()
y = (df[target].str.upper() == "THEFT").astype(int)

X_base.head()
```

```python
X_tr, X_te, y_tr, y_te = train_test_split(X_base, y, test_size=0.2, random_state=42, stratify=y)

pre = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols),
        ("num", StandardScaler(), num_cols)
    ],
    remainder="drop"
)

# Two models to compare in each version
logreg = Pipeline([("pre", pre), ("clf", LogisticRegression(max_iter=200, n_jobs=None))])
rf     = Pipeline([("pre", pre), ("clf", RandomForestClassifier(n_estimators=300, random_state=42))])
```

```python
def evaluate_model(pipe, name):
    pipe.fit(X_tr, y_tr)
    pred = pipe.predict(X_te)
    prob = pipe.predict_proba(X_te)[:,1]

    metrics = {
        "model": name,
        "accuracy": accuracy_score(y_te, pred),
        "precision": precision_score(y_te, pred),
        "recall": recall_score(y_te, pred),
        "f1": f1_score(y_te, pred),
        "roc_auc": roc_auc_score(y_te, prob),
    }

    cm = confusion_matrix(y_te, pred)
    fig, ax = plt.subplots()
    ax.imshow(cm, interpolation="nearest")
    ax.set_title(f"{name} — Confusion Matrix")
    ax.set_xticks([0,1]); ax.set_yticks([0,1])
    ax.set_xticklabels(["Neg","Pos"]); ax.set_yticklabels(["Neg","Pos"])
    for (i, j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha="center", va="center")
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    plt.show()


    RocCurveDisplay.from_predictions(y_te, prob)
    plt.title(f"{name} — ROC")
    plt.show()

    return metrics

def permutation_importances(pipe, X, y, title, n=20):

    ohe = pipe.named_steps["pre"].named_transformers_["cat"]
    num_names = num_cols
    cat_names = list(ohe.get_feature_names_out(cat_cols)) if cat_cols else []
    feat_names = cat_names + num_names

    r = permutation_importance(pipe, X, y, n_repeats=10, random_state=42, n_jobs=-1)
    imp = pd.Series(r.importances_mean[:len(feat_names)], index=feat_names).sort_values(ascending=False)[:n]

    ax = imp[::-1].plot(kind="barh")
    ax.set_title(title)
    ax.set_xlabel("Permutation importance")
    plt.tight_layout()
    plt.show()
```

```python
base_metrics = []
base_metrics.append(evaluate_model(logreg, "V1-Base: LogisticRegression"))
base_metrics.append(evaluate_model(rf,    "V1-Base: RandomForest"))
pd.DataFrame(base_metrics)
```

```python
def permutation_importances(pipe, X, y, title, n=20):

    # Transform the data using the pipeline's preprocessor
    X_transformed = pipe.named_steps["pre"].transform(X)

    ohe = pipe.named_steps["pre"].named_transformers_["cat"]
    num_names = num_cols
    cat_names = list(ohe.get_feature_names_out(cat_cols)) if cat_cols else []
    feat_names = cat_names + num_names

    # Run permutation_importance on the transformed data
    r = permutation_importance(pipe.named_steps["clf"], X_transformed, y, n_repeats=10, random_state=42, n_jobs=-1)
    imp = pd.Series(r.importances_mean, index=feat_names).sort_values(ascending=False)[:n]

    ax = imp[::-1].plot(kind="barh")
    ax.set_title(title)
    ax.set_xlabel("Permutation importance")
    plt.tight_layout()
    plt.show()

permutation_importances(logreg, X_te, y_te, "V1-Base: LogisticRegression Permutation Importances")
permutation_importances(rf, X_te, y_te, "V1-Base: RandomForest Permutation Importances")
```

The error message "Length of values (23) does not match length of index (25)" indicates that the number of feature importances calculated (23) does not match the number of feature names (25). This mismatch likely arises because the one-hot encoder might drop some categories if they are not present in the test set (X_te). To address this, we should get the feature names directly from the fitted preprocessor, which accounts for any dropped categories during the transformation.



```python
def permutation_importances(pipe, X, y, title, n=20):

    # Transform the data using the pipeline's preprocessor
    X_transformed = pipe.named_steps["pre"].transform(X)

    # Get the feature names directly from the fitted preprocessor
    feat_names = pipe.named_steps["pre"].get_feature_names_out()

    # Run permutation_importance on the transformed data
    r = permutation_importance(pipe.named_steps["clf"], X_transformed, y, n_repeats=10, random_state=42, n_jobs=-1)
    imp = pd.Series(r.importances_mean, index=feat_names).sort_values(ascending=False)[:n]

    ax = imp[::-1].plot(kind="barh")
    ax.set_title(title)
    ax.set_xlabel("Permutation importance")
    plt.tight_layout()
    plt.show()

permutation_importances(logreg, X_te, y_te, "V1-Base: LogisticRegression Permutation Importances")
permutation_importances(rf, X_te, y_te, "V1-Base: RandomForest Permutation Importances")
```

## Summary:

### Data Analysis Key Findings

*   The initial training of the Logistic Regression model failed due to a `ValueError: Input X contains NaN`, indicating that the model could not handle missing values in the input data.
*   Missing values were likely introduced in the `hour` and `dow` columns when the original "DATE\_OCC" column had missing entries.
*   The error was resolved by adding a `SimpleImputer` with a 'mean' strategy to the numerical preprocessing pipeline to handle missing values in numerical columns (`hour`, `dow`, `LAT`, `LON`) before scaling.
*   An initial error in calculating permutation importance (`ValueError: Length of values (5) does not match length of index (25)`) occurred because the importance was calculated on the original data (5 features) but feature names were derived from the one-hot encoded data (25 features).
*   The permutation importance calculation was corrected by transforming the data using the pipeline's preprocessor and obtaining the feature names directly from the fitted preprocessor using `get_feature_names_out()`.

### Insights or Next Steps

*   The imputation of missing numerical values allowed the models to train successfully.
*   Understanding feature importance after preprocessing is crucial for interpreting model results and potentially for feature selection in future iterations.


```python
# ==== SETUP ===================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, RocCurveDisplay
)
from sklearn.inspection import permutation_importance

# --- Paths (EDIT THESE) -------------------------------------------------------
LAPD_CLEAN_PATH = "/content/cleaned_property_crimes.csv"     # your cleaned LAPD file
CONTEXT_PATH    = "/content/census_context.csv"      # your context CSV (income/edu/density)

# How to join context -> LAPD. Pick one that matches your files.
JOIN_ON = "zip"             # e.g., "zip"  (preferred)
# JOIN_ON = "AREA NAME"     # or area name if that's what your context uses

# Target column helper: the code will try this list in order.
CANDIDATE_TARGETS = ["crime_binary", "crime_group", "target", "label", "is_theft"]

# ==== LOAD DATA ================================================================
base = pd.read_csv(LAPD_CLEAN_PATH)

# Try to locate the target column
target_col = None
for c in CANDIDATE_TARGETS:
    if c in base.columns:
        target_col = c
        break
if target_col is None:
    raise ValueError(
        f"Couldn't find a target column. Add one named one of {CANDIDATE_TARGETS}."
    )

# If your target is text (e.g., 'Theft'/'Break-in'), coerce to 0/1 (positive=1).
if base[target_col].dtype == "object":
    # choose the lexicographically largest as "positive" OR edit mapping to your needs
    pos = base[target_col].value_counts().index[0]
    base[target_col] = (base[target_col] == pos).astype(int)

# ==== FEATURE LISTS (robust to missing columns) ===============================
# Common useful columns you likely have; we’ll keep only those that exist.
candidate_num = ["hour", "dow", "LAT", "LON"]
candidate_cat = ["AREA NAME"]

num_cols = [c for c in candidate_num if c in base.columns]
cat_cols = [c for c in candidate_cat if c in base.columns]

if len(num_cols) + len(cat_cols) == 0:
    raise ValueError("No usable features found. Add numeric/categorical columns in the lists above.")

# ==== TRAIN/TEST SPLIT ========================================================
X = base[num_cols + cat_cols]
y = base[target_col].astype(int)
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# ==== PREPROCESSOR ============================================================
pre = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), [c for c in cat_cols if c in X_tr.columns]),
        ("num", Pipeline([("imputer", SimpleImputer(strategy="mean")),
                          ("scaler",  StandardScaler())]), [c for c in num_cols if c in X_tr.columns]),
    ],
    remainder="drop"
)

# ==== HELPERS =================================================================
def evaluate_model(pipe, name):
    pipe.fit(X_tr, y_tr)
    proba = pipe.predict_proba(X_te)[:, 1]
    pred  = (proba >= 0.5).astype(int)

    metrics = {
        "model": name,
        "accuracy": accuracy_score(y_te, pred),
        "precision": precision_score(y_te, pred, zero_division=0),
        "recall": recall_score(y_te, pred, zero_division=0),
        "f1": f1_score(y_te, pred, zero_division=0),
        "roc_auc": roc_auc_score(y_te, proba)
    }

    # Confusion Matrix
    cm = confusion_matrix(y_te, pred)
    fig, ax = plt.subplots(figsize=(4,4))
    ax.imshow(cm, interpolation="nearest")
    ax.set_title(f"{name} — Confusion Matrix")
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    for (i,j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha='center', va='center')
    plt.show()

    # ROC
    RocCurveDisplay.from_predictions(y_te, proba)
    plt.title(f"{name} — ROC")
    plt.show()

    return metrics, pipe

def permutation_importances_plot(fitted_pipe, X_data, y_data, title, top_n=20):
    # Transform with the fitted preprocessor
    X_t = fitted_pipe.named_steps["pre"].transform(X_data)

    # Get feature names AFTER preprocessing
    feat_names = fitted_pipe.named_steps["pre"].get_feature_names_out()

    # Permutation importance on the transformed data against the fitted classifier
    r = permutation_importance(
        fitted_pipe.named_steps["clf"], X_t, y_data,
        n_repeats=10, random_state=42, n_jobs=-1
    )

    imp = pd.Series(r.importances_mean, index=feat_names).sort_values(ascending=False)[:top_n]
    ax = imp[::-1].plot(kind="barh", figsize=(6, max(3, top_n*0.3)))
    ax.set_title(title)
    ax.set_xlabel("Permutation importance")
    plt.tight_layout()
    plt.show()

# ==== V1: BASELINE MODELS =====================================================
logreg_v1 = Pipeline([("pre", pre), ("clf", LogisticRegression(max_iter=200, n_jobs=None))])
rf_v1     = Pipeline([("pre", pre), ("clf", RandomForestClassifier(n_estimators=400, random_state=42, n_jobs=-1))])

v1_metrics = []
m, pipe_l1 = evaluate_model(logreg_v1, "V1-Base: LogisticRegression"); v1_metrics.append(m)
m, pipe_r1 = evaluate_model(rf_v1,     "V1-Base: RandomForest");       v1_metrics.append(m)

print(pd.DataFrame(v1_metrics))

# Feature importances (permutation) for V1
permutation_importances_plot(pipe_l1, X_te, y_te, "V1-Base: LogisticRegression — Permutation Importances", top_n=20)
permutation_importances_plot(pipe_r1, X_te, y_te, "V1-Base: RandomForest — Permutation Importances", top_n=20)

# ==== V2: CONTEXT-ENRICHED ====================================================
# Load context (must include JOIN_ON and numeric columns such as income/edu/density).
# Expect columns like: [JOIN_ON, 'median_income', 'bachelors_rate', 'pop_density']
ctx = pd.read_csv(CONTEXT_PATH)

if JOIN_ON not in base.columns or JOIN_ON not in ctx.columns:
    raise ValueError(
        f"JOIN_ON='{JOIN_ON}' must exist in BOTH files. "
        f"LAPD has: {JOIN_ON in base.columns}, context has: {JOIN_ON in ctx.columns}"
    )

# Keep only the join key + numeric context variables
context_num_cols = [c for c in ctx.columns if c != JOIN_ON and pd.api.types.is_numeric_dtype(ctx[c])]
if len(context_num_cols) == 0:
    raise ValueError("Your context CSV needs numeric columns (e.g., median_income, bachelors_rate, pop_density).")

# Merge into the **test&train universe only** to avoid leakage
lapd_full = pd.concat([X, y], axis=1)
lapd_full = lapd_full.merge(ctx[[JOIN_ON] + context_num_cols], on=JOIN_ON, how="left")

# Rebuild X/y with context
X2 = lapd_full.drop(columns=[target_col]) if target_col in lapd_full.columns else lapd_full.loc[:, X.columns.tolist() + context_num_cols]
y2 = y

# Update lists for preprocessing (keep previous + new numeric columns)
num_cols_v2 = num_cols + [c for c in context_num_cols if c not in num_cols]
cat_cols_v2 = cat_cols[:]  # same as V1

X2_tr, X2_te, y2_tr, y2_te = train_test_split(
    X2, y2, test_size=0.25, random_state=42, stratify=y2
)

pre_v2 = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), [c for c in cat_cols_v2 if c in X2_tr.columns]),
        ("num", Pipeline([("imputer", SimpleImputer(strategy="mean")),
                          ("scaler",  StandardScaler())]), [c for c in num_cols_v2 if c in X2_tr.columns]),
    ],
    remainder="drop"
)

logreg_v2 = Pipeline([("pre", pre_v2), ("clf", LogisticRegression(max_iter=200, n_jobs=None))])
rf_v2     = Pipeline([("pre", pre_v2), ("clf", RandomForestClassifier(n_estimators=400, random_state=42, n_jobs=-1))])

v2_metrics = []
# Evaluate with X2_te/y2_te to keep splits parallel
def eval_v2(pipe, name):
    pipe.fit(X2_tr, y2_tr)
    proba = pipe.predict_proba(X2_te)[:, 1]
    pred  = (proba >= 0.5).astype(int)
    metrics = {
        "model": name,
        "accuracy": accuracy_score(y2_te, pred),
        "precision": precision_score(y2_te, pred, zero_division=0),
        "recall": recall_score(y2_te, pred, zero_division=0),
        "f1": f1_score(y2_te, pred, zero_division=0),
        "roc_auc": roc_auc_score(y2_te, proba)
    }
    # Confusion matrix
    cm = confusion_matrix(y2_te, pred)
    fig, ax = plt.subplots(figsize=(4,4))
    ax.imshow(cm, interpolation="nearest")
    ax.set_title(f"{name} — Confusion Matrix")
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    for (i,j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha='center', va='center')
    plt.show()
    # ROC
    RocCurveDisplay.from_predictions(y2_te, proba)
    plt.title(f"{name} — ROC"); plt.show()
    return metrics, pipe

m, pipe_l2 = eval_v2(logreg_v2, "V2-Context: LogisticRegression"); v2_metrics.append(m)
m, pipe_r2 = eval_v2(rf_v2,     "V2-Context: RandomForest");       v2_metrics.append(m)

print(pd.DataFrame(v2_metrics))

# Feature importances (permutation) for V2
permutation_importances_plot(pipe_l2, X2_te, y2_te, "V2-Context: LogisticRegression — Permutation Importances", top_n=20)
permutation_importances_plot(pipe_r2, X2_te, y2_te, "V2-Context: RandomForest — Permutation Importances", top_n=20)

# ==== SIDE-BY-SIDE SUMMARY ====================================================
summary = pd.concat([pd.DataFrame(v1_metrics), pd.DataFrame(v2_metrics)], ignore_index=True)
display(summary.sort_values(["model"]))
```
